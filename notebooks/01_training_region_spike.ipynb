```python
import requests, time

# This is the iNaturalist API endpoint that's parameters provide us observation information
API = "https://api.inaturalist.org/v1/observations"
HEFLIN = (33.649, -85.591)           # lat, lon
MI_TO_KM = 1.609

# This function strcutures the parameters we will send to iNaturalist regarding a given species
def count_obs(**extra):
    params = {
        "taxon_name": "Terrapene carolina",   # species level, includes subspecies
        "quality_grade": "research",
        "per_page": 1,
        **extra,
    }
    r = requests.get(API, params=params, timeout=30) # get() will add the parameters to the endpoint and fetch the information (if available)
    r.raise_for_status()
    time.sleep(1)                      # be polite to the API
    return r.json()["total_results"]

results = {}
for miles in (10, 25, 50, 100):
    results[f"within {miles} mi of Heflin"] = count_obs(
        lat=HEFLIN[0], lng=HEFLIN[1], radius=miles * MI_TO_KM
    )

# Rough bounding box for Alabama
results["Alabama"] = count_obs(swlat=30.14, swlng=-88.47, nelat=35.01, nelng=-84.89)

for k, v in results.items():
    print(f"{k}: {v:,}")


extra = {
    "150 mi radius": count_obs(lat=HEFLIN[0], lng=HEFLIN[1], radius=150 * MI_TO_KM),
    "Atlanta area (east of AL box)": count_obs(swlat=33.2, swlng=-84.89, nelat=34.3, nelng=-83.9),
}
for k, v in extra.items():
    print(f"{k}: {v:,}")
```
